```python
from pathlib import Path
import sqlite3
import pandas as pd
df = pd.read_csv("data/historical_automobile_sales.csv" if Path("data/historical_automobile_sales.csv").exists() else "../data/historical_automobile_sales.csv")
con = sqlite3.connect(":memory:")
df.to_sql("automobile_sales", con, index=False, if_exists="replace")
```

```python
pd.read_sql_query("""
SELECT Recession, COUNT(*) AS months,
       ROUND(AVG(Automobile_Sales),2) AS average_sales
FROM automobile_sales GROUP BY Recession
""", con)
```

```python
pd.read_sql_query("""
SELECT Vehicle_Type, Recession,
       ROUND(AVG(Automobile_Sales),2) AS average_sales
FROM automobile_sales
GROUP BY Vehicle_Type, Recession
ORDER BY Vehicle_Type, Recession
""", con)
```

```python
pd.read_sql_query("""
SELECT City AS State, ROUND(SUM(Automobile_Sales),2) AS total_sales
FROM automobile_sales GROUP BY City ORDER BY total_sales DESC
""", con)
```

```python
pd.read_sql_query("""
SELECT Year, ROUND(SUM(Automobile_Sales),2) AS annual_sales
FROM automobile_sales GROUP BY Year ORDER BY Year
""", con)
```
