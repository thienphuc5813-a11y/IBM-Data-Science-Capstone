```python
import pandas as pd
import folium
from folium.plugins import MarkerCluster
df = pd.read_csv("../data/historical_automobile_sales.csv")
state_coords = {
 "California": (38.5767, -121.4944), "Georgia": (33.7490, -84.3880),
 "Illinois": (39.7983, -89.6544), "New York": (42.6526, -73.7562)
}
state_sales = df.groupby("City")["Automobile_Sales"].sum()
m = folium.Map(location=(39.5,-98.35), zoom_start=4, tiles="CartoDB positron")
cluster = MarkerCluster().add_to(m)
for state, sales in state_sales.items():
 folium.Marker(state_coords[state], tooltip=state,
               popup=f"{state}: {sales:,.1f} sales (dataset units)").add_to(cluster)
m
```
