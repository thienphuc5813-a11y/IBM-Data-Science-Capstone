```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
DATA = Path("../data/historical_automobile_sales.csv")
df = pd.read_csv(DATA)
df.head()
```

```python
df.info()
df.describe(include="all").T
```

```python
df["Date"] = pd.to_datetime(df["Date"])
df = df.drop_duplicates().dropna(subset=["Date", "Automobile_Sales", "Recession"])
df["Month_Num"] = df["Date"].dt.month
df["Quarter"] = df["Date"].dt.quarter
df.shape
```

```python
annual = df.groupby("Year", as_index=False)["Automobile_Sales"].sum()
sns.lineplot(data=annual, x="Year", y="Automobile_Sales", marker="o")
plt.title("Annual automobile sales"); plt.show()
```

```python
sns.barplot(data=df, x="Recession", y="Automobile_Sales", errorbar=None)
plt.title("Average monthly sales: recession vs non-recession"); plt.show()
```

```python
sns.barplot(data=df, x="Vehicle_Type", y="Automobile_Sales", hue="Recession", errorbar=None)
plt.xticks(rotation=25); plt.title("Vehicle type and recession"); plt.show()
```

```python
sns.scatterplot(data=df, x="Price", y="Automobile_Sales", hue="Recession")
plt.title("Vehicle price and sales"); plt.show()
```
